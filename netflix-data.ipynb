```python
#load the data and check shape
netflix = pd.read_csv('/kaggle/input/datasets/shivamb/netflix-shows/netflix_titles.csv')
print(netflix.shape)
print(netflix.info())
print(netflix.isnull().sum())
```

```python
# common rating 
display(netflix.head())
# set seed for reproducibility
np.random.seed(0) 
# ratings=netflix.max(rating >0)
# print{f"most common rating" ,ratings }
```

```python
# Step 1 — Fill missing values
netflix['director'] = netflix['director'].fillna('Unknown')
netflix['cast']     = netflix['cast'].fillna('Unknown')
netflix['country']  = netflix['country'].fillna('Unknown')

# Drop rows where rating or duration is null (only 4 and 3 rows — safe to drop)
netflix = netflix.dropna(subset=['rating', 'duration'])

print("Missing values after cleaning:")
print(netflix.isnull().sum())

# Step 2 — Fix date_added
netflix['date_added'] = pd.to_datetime(netflix['date_added'].str.strip())
netflix['year_added']  = netflix['date_added'].dt.year
netflix['month_added'] = netflix['date_added'].dt.month

print("\nDate range:", netflix['year_added'].min(), "to", netflix['year_added'].max())

# Step 3 — Fix duration
# Movies → extract minutes as number
# TV Shows → extract seasons as number
netflix['duration_number'] = netflix['duration'].str.extract(r'(\d+)').astype(int)

print("\nDuration sample:")
print(netflix[['type', 'duration', 'duration_number']].head(10))

# Step 4 — Explore
print("\nType distribution:")
print(netflix['type'].value_counts())

print("\nTop 10 countries:")
print(netflix['country'].value_counts().head(10))

print("\nMost common ratings:")
print(netflix['rating'].value_counts().head(10))
```
