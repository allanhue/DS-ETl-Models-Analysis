```python
# age     salary      experience    hired
# 25      50000       2             Yes
# 30      75000       5             No
# 999     120000      8             Yes
# 28      -99999      4             No
# 35      90000       150           Yes
# 22      60000       3             Yes
```

```python


df = pd.DataFrame({
    'age':        [25, 30, 999, 28, 35, 22],
    'salary':     [50000, 75000, 120000, -99999, 90000, 60000],
    'experience': [2, 5, 8, 4, 150, 3],
    'hired':      ['Yes', 'No', 'Yes', 'No', 'Yes', 'Yes']
})

print(df)

# Fix salary sentinel values
df['salary'] = df['salary'].replace(-99999, np.nan)

# Fix impossible ages — anything over 100 is bad
df['age'] = df['age'].apply(lambda x: np.nan if x > 100 else x)

# Fix impossible experience  cap at 50 years
df['experience'] = df['experience'].apply(lambda x: np.nan if x > 50 else x)

# Fill all nulls with median
for col in ['salary', 'age', 'experience']:
    median = df[col].median()
    df[col] = df[col].fillna(median) 
    print(f"{col} median used: {median}")
    
print(df)

# Detect outliers using IQR method
for col in ['salary', 'age', 'experience']:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    outliers = df[(df[col] < lower) | (df[col] > upper)]
    print(f"{col}: lower={lower:.0f}, upper={upper:.0f}, outliers={len(outliers)}")
```
